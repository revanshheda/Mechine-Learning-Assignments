```python
def test(raw_num):
    raw_num = raw_num.replace(" ", "").replace("(", "").replace(")", "").replace("-", "").replace(".", "")

    if len(raw_num) == 10 and raw_num.isdigit():
        return "(" + raw_num[:3] + ") " + raw_num[3:6] + "-" + raw_num[6:]

    return "Invalid Number"


print(test("555-987-6543"))
```
