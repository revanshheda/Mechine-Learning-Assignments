```python
def test(sku):
    sku = sku.strip().upper()

    if len(sku) != 8:
        return "Invalid SKU"

    if sku[0:3].isalpha() and sku[3] == "-" and sku[4:].isdigit():
        return "Valid: " + sku

    return "Invalid SKU"


print(test(" xyz-9876 "))
print(test("AB-12345"))
print(test("A@C-1234"))
```
